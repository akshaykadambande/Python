```python
# Marks Obtained for 10 students in a Class (Consider Marks obtained and Student Names as random)
import numpy as np
marks=np.array([98,87,90,78,75,59,85,79,69,80])
students=np.array(["Divya","Pranav","Sahil","Tanishka","Adhyay","Prerna","Sandhya","Vishal","Rahul","Smriti"])

highidx=np.argmax(marks)
lowidx=np.argmin(marks)
# 1.How many students appeared for the exam?
print("1.Total Students appeared for exam :",np.size(marks))

# 2. Calculate the total marks obtained by all students.
print("2.Total Marks obtained by all students are :",np.sum(marks))

# 3. Find Student who scored Highest Marks.
print("3.Highest Marks Obtained are : ",np.max(marks),"\nName of the Student who got Highest Marks : ",students[highidx])

# 4. Find Student who scored Lowest Marks.
print("4.Lowest Marks Obtained are : ",np.min(marks),"\nName of the Student who got Lowest Marks : ",students[lowidx])

# 5. Can you categorize the students on the basis of their marks ? 
category=np.where(marks>80,"Distinction",np.where(marks>70,"First Class",np.where(marks>60,"Pass","Fail")))
print("5.The categories of the Students as per Marks are : \n",category)

# 6. How many students passed (scored above 60)?
print("6.Total number of students who passed are :",np.sum(marks>60),"-",students[marks>60])

# 7. How many students Failed (scored below 60)?
print("7.Total number of students who Failed are :",np.sum(marks<60),"-",students[marks<60])

# 8. How many students scored above 80?
print("8.Total number of students who got Distinction are :",np.sum(marks>80),"-",students[marks>80])

# 9. How many students scored above 70?
print("9.Total number of students who got First Class are :",np.sum(marks>70),"-",students[marks>70])

# 10. Find the average (mean) marks obtained by students.
print("10.Average marks obtained by students are : ",np.mean(marks))

# 11. Find the median of marks obtained by students.
print("11.Median marks obtained by students are : ",np.median(marks))

# 12. Calculate the range of marks.
print("12.The range of marks obtained by students are : ", np.ptp(marks))

# 13. Find the variance of marks obtained by students.
print("13.Variance of marks obtained by students are : ",np.var(marks))

# 14. Calculate the pass percentage.
print("14.% of students who passed the exam are :",np.sum((marks>60)*100/10))

# 15. Calculate the fail percentage.
print("15.% of students who failed the exam are :",np.sum((marks<60)*100/10))

# 16. List the Marks of students who failed 
fail=np.where(marks<60)
print("16.The marks of the students who failed are: ",marks[fail])

# 17. Count students scoring exactly the passing marks (consider 60 as passing criteria).
barepass=np.sum(marks==60)
print("17.The students who are just passed are: ",barepass)

# 18. Find the percentage difference between pass and fail students.
print("18.The % difference between passed and filed students is: ",np.sum((marks>60)*100/10)-np.sum((marks<60)*100/10))

# 19. How many students scored above the average?
print("19.Number of students scored above average marks are: ",np.sum(marks>=np.mean(marks)))

# 20. How many students scored below the average?
print("20.Number of students scored below average marks are: ",np.sum(marks<np.mean(marks)))

# 21. Rank students based on marks.
descmarks=(np.argsort(marks)[::-1])
print("21.Ranks of students based on the marks obtained is : ",students[descmarks])

# 22. Find the Student with top three scores. 
print("22.Top three scorers are : ",students[descmarks][:3])

# 23. Find the Student with bottom three scores. 
ascmarks=np.argsort(marks)
print("23.Bottom Three scorers are : ",students[ascmarks][:3])

# 24.Which grade has the highest number of students?
unique,counts=np.unique(category,return_counts=True)
highestgrades=unique[np.argmax(counts)]
print("24.Category with Highest number of students is : ",highestgrades)
```
